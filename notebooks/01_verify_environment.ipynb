# AgniGeo Phase 1 verification
This notebook only visualizes real intermediate rasters and SOLWEIG output.

```python
from pathlib import Path
import rasterio, matplotlib.pyplot as plt
paths=list(Path('data/processed').rglob('*.tif'))
assert paths, 'First run preprocessing and real SOLWEIG.'
fig, axes=plt.subplots(1,len(paths),figsize=(5*len(paths),5))
for ax,path in zip([axes] if len(paths)==1 else axes,paths):
    with rasterio.open(path) as ds: ax.imshow(ds.read(1)); ax.set_title(path.name)
plt.show()
```
